# How factor investing works

*an illustrative example: value factor*

---

## Objective

We want to study whether **value stocks** (those with *high book-to-market ratios*) earn higher returns than **growth stocks** (low book-to-market).

We will:
1. Sort all stocks into **5 groups (portfolios)** based on their book-to-market (B/M) ratios.  
2. Compute the **average return** of each decile every month.  
3. Compute the **long-short portfolio return**:  
   $R_{LS,t} = R_{\text{P5, t}} - R_{\text{P1, t}}$
   where we are *long* in high-B/M stocks (value) and *short* in low-B/M stocks (growth).

## Step 1. What the data look like

Imagine we have monthly data for 5 stocks over two months:

| id | date       | return ($r_t$) | book-to-market ($bm_t$) |
|----|-------------|-------------|-----------------------|
| A  | 2020-01-31  | 0.03 | 0.45 |
| B  | 2020-01-31  | 0.02 | 0.80 |
| C  | 2020-01-31  | -0.01 | 0.25 |
| D  | 2020-01-31  | 0.01 | 0.50 |
| E  | 2020-01-31  | 0.04 | 0.60 |
| A  | 2020-02-29  | 0.04 | 0.47 |
| B  | 2020-02-29  | 0.01 | 0.85 |
| C  | 2020-02-29  | 0.00 | 0.26 |
| D  | 2020-02-29  | 0.03 | 0.55 |
| E  | 2020-02-29  | 0.02 | 0.58 |

## Step 2. Lag the book-to-market ratio

To avoid *look-ahead bias*, we sort each month's portfolios using **the previous month's** book-to-market ratio.

| id | date | return ($r_t$) | $bm_t$ | lagged bm ($bm_{t-1}$) |
|----|------|--------------|------|-------------------|
| A | 2020-02-29 | 0.04 | 0.47 | 0.45 |
| B | 2020-02-29 | 0.01 | 0.85 | 0.80 |
| C | 2020-02-29 | 0.00 | 0.26 | 0.25 |
| D | 2020-02-29 | 0.03 | 0.55 | 0.50 |
| E | 2020-02-29 | 0.02 | 0.58 | 0.60 |

At the end of **February**, we will use these $bm_{t-1}$ values to sort.

## Step 3. Rank the stocks by lagged B/M

| Rank | id | $bm_{t-1}$ | Interpretation |
|------|----|-------:|----------------|
| 1 | C | 0.25 | Lowest B/M → growth |
| 2 | A | 0.45 |  |
| 3 | D | 0.50 |  |
| 4 | E | 0.60 |  |
| 5 | B | 0.80 | Highest B/M → value |


## Step 4. Assign portfolios

Since we only have 5 stocks, the maximum number of portfolios we can construct is 5 (i.e. one stock in each portfolio).  
(*Note: In real research, there will be many stocks. For example, if there are 100 stocks, each portfolio will contain 20 stocks.*)

| Port | Stocks Included | Description |
|---------|-----------------|--------------|
| 1 | C | Lowest B/M (growth) |
| 2 | A | Low B/M |
| 3 | D | Medium |
| 4 | E | High B/M |
| 5 | B | Highest B/M (value) |


## Step 5. Compute portfolio returns (equal-weighted)

Each stock's **return this month ($r_t$)** is the realized return during February.  
We simply take the *average return* within each decile.

| Port | Stocks | Returns ($r_t$) | Average Return |
|---------|---------|--------------|----------------|
| 1 | C | 0.00 | 0.00 |
| 2 | A | 0.04 | 0.04 |
| 3 | D | 0.03 | 0.03 |
| 4 | E | 0.02 | 0.02 |
| 5 | B | 0.01 | 0.01 |


## Step 6. Compute the long-short return

We go **long** portfolio 5 (high B/M) and **short** portfolio 1 (low B/M):

$R_{LS, Feb} = R_{P5,Feb} - R_{P1,Feb} = 0.01 - 0.00 = 0.01$

The long-short portfolio earns **1%** in February.

## Step 7. Repeat for each month

You repeat this process for each calendar month.

| Month | Return (port 1) | Return (port 5) | Long-Short (5-1) |
|--------|------------------:|-------------------:|------------------:|
| Feb-20 | 0.00 | 0.01 | 0.01 |
| Mar-20 | 0.02 | 0.04 | 0.02 |
| Apr-20 | -0.01 | 0.00 | 0.01 |
| ... | ... | ... | ... |


## Step 8. Compute average and t-statistic

After you have a time series of monthly long-short returns, you can summarize it:

| Statistic | Formula | Example Value |
|------------|-------|---------------|
| Mean monthly return | $$\bar{R} = \frac{1}{T}\sum R_t$$ | 0.012 |
| Standard deviation | $$s = [\frac{1}{T-1}\sum (R_t - \bar{R})^2]^{1/2}$$ | 0.015 |
| Standard error | $$\frac{s}{T^{1/2}}$$ | 0.003 |
| t-statistic | $$\frac{\bar{R}}{s/(T^{1/2})}$$ | 4.0 |
| Annualized mean | $$12 \times \bar{R}$$ | 0.144 |
| Annualized volatility | $$12^{1/2} \times s$$ | 0.052 |

Interpretation:  
If the **t-statistic > 2 (or 1.96)**, the long-short average return is significantly different from zero - evidence that *value stocks outperform growth stocks*.

## Key Lessons

1. **Always lag your sorting variable** (use $bm_{t-1}$, not $bm_t$).  
   → This mimics real-time decision making and avoids using future info.

2. **Sorting is cross-sectional:** each month, you form groups across all stocks.  

3. **Portfolio returns are time-series:** each month you compute decile returns and track them over time.  

4. **Equal-weighted vs Value-weighted:**  
   - Equal-weighted → every firm counts the same.  
   - Value-weighted → large firms have more impact.

5. **Long-short interpretation:**  
   - If positive → value (high B/M) outperform growth (low B/M).  
   - If negative → growth outperform value.

6. **Extensions for practice:**  
   - Try 10 portfolios (deciles) instead of 5.  
   - Sort on other firm characteristics (momentum, size, profitability).  
   - Compare equal-weighted and value-weighted results.  
   - Plot cumulative performance of the long-short strategy.

**Summary**

| Step | What You Do | Why |
|------|--------------|-----|
| 1 | Prepare panel data | Organize inputs |
| 2 | Lag B/M ratio | Prevent look-ahead bias |
| 3 | Rank and assign deciles each month | Form portfolios |
| 4 | Compute portfolio returns | Measure performance |
| 5 | Take long - short | Construct long-short |
| 6 | Average over time | Estimate expected premium |
| 7 | Evaluate t-stat | Test statistical significance |

This is the workflow how a cross-sectional sort becomes a time-series portfolio strategy - from raw data → deciles → returns → long-short factor.
